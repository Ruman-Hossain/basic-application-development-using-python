# Maintain The Record of Performed JOb

## Required Library Packages Installation

```python
import ipywidgets as widgets
from IPython.display import display
import pandas as pd

# Initialize an empty DataFrame to store the data
data = pd.DataFrame(columns=["Name", "Age"])
```

## Function To Handle Form Submission

```python
def save_data(b):
    name = name_input.value
    age = age_input.value

    if name and age:
        data.loc[len(data)] = [name, age]
        print("Data saved successfully.")
```

## Function to handle search

```python
def search_data(b):
    name = name_input.value
    result = data[data["Name"] == name]

    if not result.empty:
        print("Search results:")
        print(result)
    else:
        print("No matching records found.")
```

## Create input widgets

```python
name_input = widgets.Text(description="Name:")
age_input = widgets.Text(description="Age:")
```

## Create buttons

```python
save_button = widgets.Button(description="Save")
search_button = widgets.Button(description="Search")
```

## For save & search data

```python
save_button.on_click(save_data)
search_button.on_click(search_data)
```

## Display widgets

```python
display(name_input)
display(age_input)
display(save_button)
display(search_button)
```
