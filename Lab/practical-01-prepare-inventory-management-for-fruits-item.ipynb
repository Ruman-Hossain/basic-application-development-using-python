## Take Item Data from the User (Like Fruit Name, Unit Price, Quantity and Total Price Etc)

```python
def get_fruit_data():
  fruit_name = input("Enter Fruit Name: ")
  unit_price = float(input("Enter Unit Price: "))
  quantity = int(input("Enter Quantity: "))
  total_price = unit_price * quantity

  return{
      "fruit_name": fruit_name,
      "unit_price": unit_price,
      "quantity": quantity,
      "total_price": total_price
  }
```

## Store Item Data Into CSV File

```python
import csv
import os

def store_data_to_csv(data, filename="fruits_inventory.csv"):
  field_names = ["fruit_name","unit_price","quantity","total_price"]
  file_exists = os.path.exists(filename)

  with open(filename, 'a', newline='') as csvfile:
    writer = csv.DictWriter(csvfile, fieldnames = field_names)

    if not file_exists or csvfile.tell() == 0: # Check if file is new or empty
      writer.writeheader() # Write header only if file is new or empty

    writer.writerow(data)
```

## Print Fruit Items Summary Data From Stored CSV File



```python
import csv
import os

def print_fruits_summary(filename="fruits_inventory.csv"):
  if not os.path.exists(filename) or os.stat(filename).st_size == 0:
    print("No fruit items to display. The inventory file is empty or does not exist.")
    return

  with open(filename, 'r', newline='') as csvfile:
    reader = csv.DictReader(csvfile)

    # Check if the CSV has headers after opening
    if not reader.fieldnames:
      print("The inventory file exists but has no headers. Please ensure the file is correctly formatted.")
      return

    try:
      print(f"{'Fruit Name':<15}{'Unit Price':<15}{'Quantity':<10}{'Total Price'}")
      print('-'*50)

      found_data = False
      for row in reader:
        found_data = True
        print(f"{row['fruit_name']:<15}{float(row['unit_price']):<15}{int(row['quantity']):<10}{float(row['total_price']):.2f}")

      if not found_data:
        print("The inventory file has headers but no data rows.")

    except KeyError as e:
      print(f"Error reading fruit data: Missing expected column '{e}' in the CSV file.")
    except Exception as e:
      print(f"An unexpected error occurred: {e}")
```

## Maintain the Record of the Performed Job

```python
import datetime

def record_job(log_filename="job_log.txt", job_description=""):
  timestamp = datetime.datetime.now().strftime("%Y-%m-%d %H:%M:%S")

  with open(log_filename, 'a') as logfile:
    logfile.write(f"[{timestamp}]{job_description}\n")
```

## Main Program

```python
while True:
  print("\nInventory Management")
  print("1. Add New Fruit Item")
  print("2. View Fruit Items Summary")
  print("3. Exit")
  choice = input("Choose an Option: ")

  if choice == '1':
    #Take Item Data From The User
    data = get_fruit_data()

    #Store Items Data Into CSV File
    store_data_to_csv(data)

    #Maintain The Record Of Performed JOb
    record_job(job_description=f"Addedd Fruit Item: {data}")

    print("Fruit Item Added Successfully!")

  elif choice == '2':
    #Print Fruits Items Summary Data From Stored CSV File
    print_fruits_summary()

  elif choice =='3':
    print("Exiting The Program.")
    break

  else:
    print("Invalid Choice, Please Try Again")
```
