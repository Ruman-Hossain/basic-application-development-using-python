## In The Report Table Show Students Basic Information

```python
import pandas as pd
import csv
import datetime
from reportlab.lib.pagesizes import letter
from reportlab.pdfgen import canvas

# Take item data from the user
def get_student_data():
    ID = input("Enter Student ID: ")
    Name = input("Enter Student Name: ")
    Roll = input("Enter Roll: ")
    Semester = input("Enter Semester: ")
    Shift = input("Enter Shift: ")
    Department = input("Enter Department Name: ")
    Date_Of_Birth = input("Enter your DOB (Day-Month-year): ")
    Phone_number = input("Enter Your Phone number: ")

    return {
        "ID": ID,
        "Name": Name,
        "Roll": Roll,
        "Semester": Semester,
        "Shift": Shift,
        "Department": Department,
        "Date_Of_Birth": Date_Of_Birth,
        "Phone_number": Phone_number
    }
```

## Store data into CSV file


```python
def store_data_to_csv(data, filename="students_info.csv"):
    fieldnames = ["ID", "Name", "Roll", "Semester", "Shift", "Department", "Date_Of_Birth", "Phone_number"]

    with open(filename, 'a', newline="") as csvfile:
        writer = csv.DictWriter(csvfile, fieldnames=fieldnames)
        # Write the data
        writer.writerow(data)
```

## For generating PDF report

```python
from reportlab.lib.pagesizes import letter
from reportlab.pdfgen import canvas
from reportlab.lib.units import inch
import pandas as pd # Ensure pandas is imported if not already in this cell

def generate_pdf_report(filename="students_info.csv", pdf_filename="students_report.pdf"):
    # Create a new PDF file
    c = canvas.Canvas(pdf_filename, pagesize=letter)
    width, height = letter

    # Define text positions and line height
    x_start = 72  # Starting X position
    y_start = height - 1 * inch  # Starting Y position
    line_height = 18 # Height for each line of text

    # Read CSV file
    try:
        data_frame = pd.read_csv(filename)
    except FileNotFoundError:
        print(f"Error: The file '{filename}' was not found. No report generated.")
        return
    except pd.errors.EmptyDataError:
        print(f"Error: The file '{filename}' is empty. No report generated.")
        return

    # Add a title to the PDF
    c.setFont('Helvetica-Bold', 14)
    c.drawString(x_start, y_start, 'Student Information Report')
    y_start -= 0.5 * inch # Move down after title

    # Add column headers
    c.setFont('Helvetica-Bold', 10)
    headers = data_frame.columns.tolist()
    header_line = ' | '.join(headers)
    c.drawString(x_start, y_start, header_line)
    y_start -= line_height # Move down after headers

    # Draw a line separator
    c.line(x_start, y_start, width - x_start, y_start)
    y_start -= line_height / 2

    # Set font for data rows
    c.setFont('Helvetica', 9)

    # Insert data from each row
    for index, row in data_frame.iterrows():
        if y_start < (1 * inch): # Check if enough space for next row, if not, new page
            c.showPage() # Start a new page
            y_start = height - 1 * inch # Reset y_start for new page
            c.setFont('Helvetica-Bold', 10)
            c.drawString(x_start, y_start, header_line) # Redraw headers on new page
            y_start -= line_height
            c.line(x_start, y_start, width - x_start, y_start)
            y_start -= line_height / 2
            c.setFont('Helvetica', 9)

        # Format data for display (simple string join for now)
        row_data = [str(row[col]) for col in headers]
        data_line = ' | '.join(row_data)
        c.drawString(x_start, y_start, data_line)
        y_start -= line_height # Move down for the next row

    # Save the PDF
    c.save()
    print(f"PDF report generated: {pdf_filename}")
```

## Main program

```python

def main():
    while True:
        print("\nStudent Info Management")
        print("1. Add new student info")
        print("2. Generate PDF")
        print("3. Exit")

        choice = input("Choose an option: ")

        if choice == '1':
            # Take item data from the user
            data = get_student_data()

            # Store student info into CSV file
            store_data_to_csv(data)
            print("Student data added successfully!")

        elif choice == '2':
            # Print PDF report
            generate_pdf_report()

        elif choice == '3':
            print("Exiting the program.")
            break
        else:
            print("Invalid choice. Please try again.")
```

# Run The Main Program

```python
if __name__ == "__main__":
    main()
```
